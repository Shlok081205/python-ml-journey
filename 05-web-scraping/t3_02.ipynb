```python
"""
parse_local_html.py
--------------------
Parses the locally saved Screener HTML file and produces a CSV.
Use this if you've already downloaded the pages, or as a sanity-check
before running the full live scraper.
"""

from bs4 import BeautifulSoup as bs
import pandas as pd

COLUMNS = [
    "S.No.", "Link", "Company",
    "CMP (Rs.)", "P/E", "Mar Cap (Rs.Cr.)",
    "Div Yld (%)", "NP Qtr (Rs.Cr.)", "Qtr Profit Var (%)",
    "Sales Qtr (Rs.Cr.)", "Qtr Sales Var (%)", "ROCE (%)",
]


def parse_html_file(path: str) -> pd.DataFrame:
    with open(path, encoding="utf-8") as f:
        soup = bs(f, "html.parser")

    records = []
    for row in soup.find_all("tr"):
        # Skip header rows (they use <th>, not <td>)
        if not row.get("data-row-company-id"):
            continue

        tds = row.find_all("td")
        if len(tds) < 3:
            continue

        sno     = tds[0].get_text(strip=True)
        a_tag   = tds[1].find("a")
        if not a_tag:
            continue

        href    = a_tag.get("href", "")
        company = a_tag.get_text(strip=True)
        nums    = [td.get_text(strip=True) for td in tds[2:]]

        records.append([sno, href, company] + nums)

    return pd.DataFrame(records, columns=COLUMNS)


if __name__ == "__main__":
    import sys
    path = sys.argv[1] if len(sys.argv) > 1 else "All_Listed_Companies_-_Screener.html"
    df = parse_html_file(path)
    print(f"Parsed {len(df)} rows from {path}")
    print(df.head())
    out = path.replace(".html", ".csv")
    df.to_csv(out, index=False)
    print(f"Saved → {out}")
```
